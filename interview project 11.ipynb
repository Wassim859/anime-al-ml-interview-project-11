```python

tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(
    df["combined_features"]
)

cosin_sim = cosine_similarity(tfidf_matrix)

print("Similarity matrix created successfully!")
print("Similarity matrix shape:", cosin_sim.shape)
```

```python
cosine_sim = cosine_similarity(tfidf_matrix)
print("Cosine Similarity Shape:")
print(cosine_sim.shape)
```

```python
indices = pd.Series(df.name,index=df["name"]).drop_duplicates()
```

```python
def recommend_anime(anime_name, num_recommendation=5):

    if anime_name not in indices.index:
        return "Anime not found in dataset."
    anime_index = int(indices.loc[anime_name])

    similarity_scores = list(
        enumerate(cosin_sim[anime_index])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = [
        item for item in similarity_scores
        if item[0] != anime_index
    ]

    similarity_scores = similarity_scores[
        :num_recommendation
    ]

    anime_indices = [i[0] for i in similarity_scores]

    result = df.iloc[anime_indices][
        ["name", "genre", "type", "rating"]
    ].copy()

    result["similarity_score"] = [
        round(i[1], 3)
        for i in similarity_scores
    ]

    return result.reset_index(drop=True)
```
